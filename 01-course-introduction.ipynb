# How the class works

**Four classes of three hours**, each including a break.

We move from algorithms on paper to Python, then from functions and collections to object-oriented programming. Classes 2–4 combine short demonstrations with increasingly substantial implementation tasks.

1. Predict and trace a worked example.
2. Implement the core task and record tests.
3. Compare solutions and explain their design.
4. Continue with a challenge when the core task is complete.

By the end, you should be able to design an algorithm, implement and test functions, process collections of records, and build cooperating Python objects.

# Learning habits

- Predict what a short example will do before running it.
- Trace values on paper when an algorithm is confusing.
- Test ordinary inputs and boundaries such as zero, equal values, or empty collections.
- Treat mistakes as evidence: explain what happened and change one thing at a time.
- Ask for help in class and explain what you have tried.
- Keep your flowcharts; we will reuse them when writing Python.

# Four-class plan

Class 1 establishes computational thinking and Python basics. Class 2 develops control flow and text processing. Class 3 combines functions and structured data in a gradebook. Class 4 starts object-oriented programming and rebuilds the gradebook with classes.

Notebooks **01–09** follow the teaching order. Notebook **08** contains the class 3 gradebook project and an additional algorithm practice bank. Notebook **09** is core class 4 material.

Complete the core exercises before choosing further challenges. Repeated maximum and counting tasks serve as quick checks; the integrated projects are the main work. Detailed scheduling is in the [README](README.md).
