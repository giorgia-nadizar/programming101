# How the class works

**Four classes of three hours**, each including a break.

We begin with algorithms on paper, then translate familiar ideas into Python. No previous programming knowledge is assumed.

1. Predict or discuss a small problem.
2. Study a short worked example.
3. Practise individually or in pairs; alternate designer and executor.
4. Test, compare, and improve your solution.

By the end, you should be able to specify a problem, draw a flowchart, write a Python solution using functions and collections, and explain how you tested it.

# Learning habits

- Predict what a short example will do before running it.
- Trace values on paper when an algorithm is confusing.
- Test ordinary inputs and boundaries such as zero, equal values, or empty collections.
- Treat mistakes as evidence: explain what happened and change one thing at a time.
- Ask for help in class and explain what you have tried.
- Keep your flowcharts; we will reuse them when writing Python.

# Four-class plan

The course meets **four times for three hours**, with a break in each class. Exercises, discussion, and review are included; no homework is required to keep pace.

The first class starts with introductions, followed by computational thinking and then Python. Students need only paper and pens for the computational-thinking activities. Have a working Jupyter environment available before starting Python.

Notebooks **01–08** follow the teaching order, with lists and dictionaries before the implementation workshop. Notebook **09** is an optional extension. [Classes and objects](09-classes-and-objects-optional.ipynb), list comprehensions, and two-dimensional averages are optional follow-up topics, outside the scheduled core. If the group needs more time, prioritise the main exercise in each block and shorten the extra challenge or whole-class sharing.
