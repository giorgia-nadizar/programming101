## 1. Computational thinking

Computational thinking is a way to formulate problems and design solutions that can be followed systematically by a person or a machine.

Imagine organising a class picnic:

| Idea | Meaning | Picnic example |
| --- | --- | --- |
| Decomposition | Split a problem into manageable parts. | Count participants, calculate food quantities, plan transport. |
| Pattern recognition | Notice similar tasks or repeated structure. | Each participant needs the same number of sandwiches. |
| Abstraction | Keep the information relevant to the question. | To count sandwiches, use participant count and sandwiches per person; ignore clothing colour. |
| Algorithm design | Give precise, ordered steps to produce a result. | Read the two counts, multiply them, report the total. |

Discuss: what extra information would matter if some people have dietary restrictions? Abstraction depends on the problem you want to solve.

An **algorithm** is a precise procedure. For the finite tasks in this lesson, it must finish after a finite number of steps for every allowed input and produce the specified output. A **program** expresses a procedure in a programming language; an algorithm can first be written in everyday language or drawn.

### Specify the problem before the steps

For the picnic calculation:

- **Inputs:** number of people and sandwiches per person.
- **Assumptions:** both inputs are nonnegative whole numbers; everyone receives the same quantity.
- **Output:** total number of sandwiches.
- **Steps:** read the inputs, multiply them, report the result.
- **Tests:** 8 people and 2 each gives 16; 0 people gives 0.

A correct solution to a poorly specified problem may still be unhelpful. Ask what is allowed and what the answer should mean.

## 2. The language of block diagrams

A flowchart shows the order of actions and the paths an execution can take.

| Shape | Meaning | Example text inside |
| --- | --- | --- |
| Oval / rounded terminator | Start or end | Start; End |
| Parallelogram | Input or output | Read height; Display total |
| Rectangle | Processing or updating a value | total ← total + value |
| Diamond | Decision with labelled outgoing arrows | Is total greater than 10? |
| Arrow | Next step / direction of control | Follow the arrow to the next block |

Label the two decision exits **Yes** and **No**. Connect every path; a loop has an arrow returning to an earlier decision. Use short, unambiguous instructions in each block.

A **variable** is a named place to keep a value. We use `←` to mean “store the value on the right in the name on the left.” For example, `count ← count + 1` reads the old count and stores a value one larger. This is an update, not a mathematical equality.

Three building blocks:

- **Sequence:** perform actions in order.
- **Selection:** choose a path using a condition.
- **Repetition:** return to a step while a condition holds. Identify what changes and why the repetition will end.

## Worked example A: sequence — triangle area

**Inputs:** positive base and height in the same unit. **Output:** area in square units. Assume valid input; validation is a separate task.



Trace with base 4 and height 10: after the calculation, area is 20; the output is 20. Repeat with base 3 and height 5: the output is 7.5. Keep fractional results.

## Worked example B: decision and repetition — countdown

**Input:** none; the starting count is fixed at 3. **Output:** 3, then 2, then 1.



| Count at decision | Is count > 0? | Output this time | Count after update |
| ---: | --- | --- | ---: |
| 3 | Yes | 3 | 2 |
| 2 | Yes | 2 | 1 |
| 1 | Yes | 1 | 0 |
| 0 | No | Nothing | No update |

The condition is checked **before** the body. The count decreases toward zero, so the loop stops. Predict what happens if the initial count is 0, or if the update block is removed.

## Exercise 1 — The literal robot

Work in pairs: read, design, execute and revise, then discuss.

Draw a 4 × 4 grid. Number rows 1–4 from top to bottom and columns 1–4 from left to right. A robot starts at (row 4, column 1), facing up. The goal is (row 1, column 4). There are no obstacles.

The robot understands only **move forward one square**, **turn left 90°**, and **turn right 90°**. It cannot leave the grid. Turning does not move it.

1. Split the route into smaller parts and write exact instructions.
2. Draw a sequence flowchart from Start to End.
3. Give it to your partner, who must execute it literally, tracking position and direction.
4. Revise any ambiguous instructions. What repeated pattern could be shortened using repetition?

**Hand in:** a diagram and a position/direction trace. The final position must be (1, 4); final direction is unrestricted. Does the same algorithm work from every starting position? Explain which assumptions it uses.

## Exercise 2 — Maximum of two numbers

Specify the problem, draw and trace your algorithm, then compare solutions.

**Inputs:** two numbers, a and b; negative values and ties are allowed. **Output:** the larger numerical value, displayed once. For a tie, display that shared value once.

Draw a flowchart with input, a decision, output, and an end. Do not use a built-in “maximum” operation: show the comparison and branches yourself.

| Inputs (a, b) | Expected output | Branch followed | Your output |
| --- | ---: | --- | --- |
| (8, 3) | 8 | | |
| (2, 9) | 9 | | |
| (4, 4) | 4 | | |
| (−5, −2) | −2 | | |

**Discuss:** do you need a separate branch for equality? Can two different diagrams both be correct? Explain why using tests and the meaning of the branches.

## Exercise 3 — Running total

Specify the values you need to remember, draw the algorithm, trace it, and revise.

**Input:** a nonnegative whole number N. **Output:** the sum of all whole numbers from 1 through N, including N. When N is 0, output 0. Assume valid input.

Design a flowchart that adds one number at a time. Do not use the closed-form formula. Your diagram needs a running total, a counter, a decision, an update, and a return arrow.

1. Choose initial values for the total and counter.
2. Decide when another number should be added.
3. Show how both values change and where the output belongs.
4. Explain why the loop ends for every allowed N.

**Tests:** N = 0 → 0; N = 1 → 1; N = 4 → 10.

Trace N = 4, including the final failed condition:

| Counter before test | Total before test | Condition result | Total after body | Counter after body |

| | | | | |
| | | | | |
| | | | | |
| | | | | |

**Stretch:** change the diagram to sum only even numbers from 1 through N.

## Exercise 4 — Debug a loop

This procedure should display 1, 2, 3, then stop:

1. Start.
2. Store 1 in count.
3. Is count ≤ 3? If No, go to End. If Yes, continue.
4. Display count.
5. Return to step 3.
6. End.

Trace the algorithm, fix the diagram, and explain the fix.

- Trace the first four visits to the decision. What is displayed? Why does this procedure fail to finish?
- Draw a corrected flowchart. Mark the missing update and put it in the correct place.
- If the condition becomes count < 3, what changes?
- If the update happens before displaying, what values appear?

**Hand in:** a corrected diagram plus expected output for each version. Trace carefully: a tiny change can produce a different result.

## Exchange diagrams and debrief

Swap your solution to exercise 2 or 3 with another pair. Execute the diagram, then discuss feedback. The original designers must stay silent during execution.

- Are inputs, allowed values, and outputs explicit?
- Are actions precise and arrows connected?
- Does each decision have labelled Yes and No paths?
- Are variables initialised before use?
- Does each loop make progress and eventually stop?
- Does the output match a normal case and a boundary case?

Record one counterexample or ambiguity, then improve the diagram. Passing a few tests builds confidence; explain why the steps also work for the other allowed inputs.

## Exit ticket and bridge to Python

Individually:

1. Explain algorithm versus program in one sentence.
2. Name one piece of information you could ignore in a problem, and why.
3. Explain what makes your running-total loop stop.
4. Give a test that catches a missed equality or boundary case.

Later, the input/output blocks will become input and output instructions; process blocks will become calculations and assignments; diamonds will become conditions; return arrows will become loops. Functions will let us name and reuse a procedure.

Keep your diagrams. In [Python basics](03-python-basics.ipynb), begin by translating the triangle-area example. In the control-flow lesson, return to maximum-of-two and running total.

## Additional in-class exercises

These are optional alternatives or revision activities. For each, submit a specification, flowchart, and at least three traces.

### Exercise 5 — Ticket price

Input is age, assumed to be a nonnegative whole number. Output a ticket price: under 6 pays €0; ages 6 through 17 pay €5; ages 18 through 64 pay €10; ages 65 and above pay €5. Draw the decisions and label all branches. Test ages 5, 6, 17, 18, 64, and 65. Could you simplify the diagram without changing its outputs?

### Exercise 6 — Largest of three

Read three numbers and output the largest value once. Ties and negative inputs are allowed. Decompose the task into two uses of your maximum-of-two procedure. Draw those steps as named process blocks, and point to the earlier diagram that defines each one. Test (3, 8, 5), (7, 7, 2), and (−4, −9, −2). Explain how this anticipates a reusable function.

### Exercise 7 — Search a row of cards

Input is a finite row of number cards and a target number. Output the position of the **first** matching card, counting positions from 1, or “not found”. The row may be empty. Examine cards from left to right, one at a time.

Draw a flowchart that checks whether a card remains before reading it. Stop when a match is found or the row is exhausted. Test an empty row, a match on the first card, duplicate matches, and an absent target. How many cards do you inspect in the best and worst cases for a row of length N?

## Instructor debrief notes — reveal after students attempt the tasks

The notes below are suggested solutions, not the only valid diagrams. Keep this section out of view while students work.

### Exercise 1

One route: move forward three times, turn right, move forward three times. The trace is (4,1,up) → (3,1,up) → (2,1,up) → (1,1,up) → (1,1,right) → (1,2,right) → (1,3,right) → (1,4,right). Repetition can replace each group of three identical moves. The fixed route relies on the specified starting state.

### Exercise 2

Start → Read a, b → Is a ≥ b? Yes → Display a; No → Display b. Both branches lead to End. Equality is handled by Yes; using a > b instead also works, with ties on No. Expected outputs are 8, 9, 4, −2. Watch for a missing equality path or displaying both values.

### Exercise 3

Start → Read N → total ← 0 → counter ← 1 → Is counter ≤ N?

- Yes → total ← total + counter → counter ← counter + 1 → return to decision.
- No → Display total → End.

For N = 4:

| Counter before test | Total before test | Condition | Total after body | Counter after body |
| ---: | ---: | --- | ---: | ---: |
| 1 | 0 | Yes | 1 | 2 |
| 2 | 1 | Yes | 3 | 3 |
| 3 | 3 | Yes | 6 | 4 |
| 4 | 6 | Yes | 10 | 5 |
| 5 | 10 | No | — | — |

Before every test, total is the sum of the numbers from 1 up to counter − 1. The counter increases by 1, so it eventually exceeds finite N. For N = 0 the first test is false and the output is 0. For the even-only extension, start counter at 2 and increase by 2.

### Exercise 4

The faulty version displays 1 forever. Add count ← count + 1 after the display and before returning to the decision. The corrected version displays 1, 2, 3. With the corrected update and condition count < 3, it displays 1, 2. With the original condition count ≤ 3 but the update before the display, it displays 2, 3, 4.

### Extensions

- **Ticket:** test age < 6, then age < 18, then age < 65; prices are 0, 5, 10, 5 on the respective paths. Boundary-test outputs: 0, 5, 5, 10, 10, 5.
- **Largest of three:** intermediate ← maximum(a, b); result ← maximum(intermediate, c); display result. Test outputs: 8, 7, −2.
- **Search:** position ← 1; while position ≤ number of cards, compare that card to target. A match displays position and ends; otherwise increment position and return to the bounds check. Exhaustion displays “not found”. For a nonempty row the best case examines 1 card; the worst examines N. An empty row examines 0. For cards [4, 2, 4] with target 4, output 1, not 3.
