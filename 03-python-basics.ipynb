## 2. Values, expressions, and types

An expression combines values and operations to produce a value. A value has a type, which determines which operations make sense.

| Type | Example | Meaning |
| --- | --- | --- |
| `int` | `5` | Whole number |
| `float` | `2.5` | Floating-point number; many decimal values are approximate |
| `bool` | `True` | A truth value: `True` or `False` |
| `str` | `"hello"` | Text |
| `NoneType` | `None` | Absence of a value |

Use `type(value)` to inspect a type. We will introduce collections when we need them in class 3.

````python
print(type(5))
print(type(5.0))
print(type("5"))
print(type(True))
````

### Numeric operations

| Operation | Python | Example result |
| --- | --- | --- |
| Add, subtract, multiply | `+`, `-`, `*` | `3 * 4` gives `12` |
| Divide | `/` | `3 / 2` gives `1.5` |
| Floor division | `//` | `7 // 2` gives `3` |
| Remainder | `%` | `7 % 2` gives `1` |
| Power | `**` | `2 ** 3` gives `8` |

Parentheses make the intended order explicit. Predict the three results before running.

````python
print(3 * 4 - 2)
print((3 * 4) - 2)
print(3 * (4 - 2))
````

## 3. Variables remember values

`name = expression` evaluates the expression and assigns its value to a name. This is the same update that we wrote as `←` in a flowchart. It is not an equality test.

Choose descriptive names such as `people` and `sandwiches_per_person`. In Python, use underscores instead of spaces.

Trace the values below. Does changing `people` automatically recalculate `total`?

````python
people = 8
sandwiches_per_person = 2
total = people * sandwiches_per_person
print(total)
people = 10
print(total)
total = people * sandwiches_per_person
print(total)
````

### Converting values

`int`, `float`, and `str` can convert suitable values. Converting a float to an integer truncates toward zero; it does not round to the nearest integer. Converting unsuitable text such as `"hello"` to an integer raises an error.

````python
print(int(7.9))
print(float("2.5"))
print(str(12))
````

## 4. Worked example — area of a circle

**Input:** a positive radius. **Output:** the circle’s area. Assume the radius uses one consistent unit.

Sequence: choose a radius → calculate π × radius² → display area. We use an approximate value of π here.

Match each Python line to a flowchart block. Change the radius to 1 and predict the result.

````python
pi = 3.14159
radius = 2.2
area = pi * radius ** 2
print(area)
````
