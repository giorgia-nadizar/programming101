## 1. Comparisons answer questions

A comparison produces `True` or `False`.

| Operator | Question |
| --- | --- |
| `>` / `<` | Greater / smaller? |
| `>=` / `<=` | Greater or equal / smaller or equal? |
| `==` | Equal? |
| `!=` | Different? |

`=` assigns a value; `==` compares values. Compare values of suitable types: `3` and `"3"` represent different kinds of data.

````python
a = 3
b = 5
print(a < b)
print(a == b)
print(a != b)
print(a == "3")
````

### Combine conditions

`and` requires both conditions to be true; `or` requires at least one; `not` reverses a truth value. Predict what changes when `has_ticket` becomes `False`.

````python
age = 20
has_ticket = True
print(age >= 18 and has_ticket)
print(age < 6 or age >= 65)
print(not has_ticket)
````

## 3. A while loop repeats while a condition is true

Return to the countdown diagram. Initialise the state, check the condition, execute the body, update the state, then check again.

Predict the output. Identify the line that ensures progress toward stopping.

````python
count = 3
while count > 0:
    print(count)
    count = count - 1
````

### Trace the final failed condition (exit condition) too

## 4. A for loop visits items in order

Use `for` when you want to visit each item in a collection or a range of integers. Use `while` when repetition is most naturally controlled by a condition. Choose the form that makes the algorithm clearest.

`range(start, stop, step)` excludes `stop`. Defaults are start 0 and step 1; step cannot be zero. A positive step moves upward and a negative step downward.

Predict every output before running:

````python
print("range(3):")
for number in range(3):
    print(number)

print("range(2, 7, 2):")
for number in range(2, 7, 2):
    print(number)

print("countdown:")
for number in range(3, 0, -1):
    print(number)
````

## Beyond single loops — search and early stopping

**Core.** A loop can stop before exhausting all candidates. `break` exits the nearest enclosing loop. Keep a flag when code after the loop needs to know whether a match was found.

The example tests whether a whole number is prime. A prime is at least 2 and has no positive divisors other than 1 and itself. Trace 2, 9, and 11. For 9, which candidates are actually visited?

````python
number = 9
is_prime = number >= 2
for divisor in range(2, number):
    if number % divisor == 0:
        is_prime = False
        break
print(is_prime)
````
