## 1. Strings store text

A string is an ordered sequence of characters, including spaces and punctuation. Quotes mark its boundaries; they are not part of its content.

`+` joins strings, `*` repeats them, and `len` counts characters.

````python
greeting = "Hello"
name = "Ada"
message = greeting + " " + name
print(message)
print(len(message))
print("ha" * 3)
````

## 2. Indexing and slicing

Indexing starts at 0. A negative index counts from the end: −1 is the last character. A slice `text[start:stop]` excludes the stop position.

For `"Python"`, indices 0 through 5 are valid. The empty string has length 0 and has no valid character index.

````python
text = "Python"
print(text[0])
print(text[-1])
print(text[1:4])
print(text[:2])
print(text[2:])
print(text[::-1])
````

### Strings are immutable

You cannot replace a character by assigning to an index. You can construct a new string and assign it to a variable. Methods such as `lower` and `upper` return new strings.

Predict whether `text` changes below.

````python
text = "Hello"
changed = "Y" + text[1:]
print(text)
print(changed)
print(text.lower())
print(text)
````

## 3. Worked example — Count spaces

**Input:** a string. **Output:** its number of space characters. An empty string gives 0.

The algorithm starts a counter at zero, visits every character, and increments only when the character matches. Follow one iteration with a non-space character and one with a space.

````python
text = "we learn Python"
spaces = 0
for character in text:
    if character == " ":
        spaces = spaces + 1
print(spaces)
````

## Text processing — normalise before comparing

**Core.** Real inputs often contain extra spaces or inconsistent capitalisation. `strip()` removes surrounding whitespace; `lower()` returns lowercase text. `split(separator)` produces a list of pieces; for now, unpack a known number of pieces into separate names. We study lists fully in class 3.

An f-string inserts expression values into text. Predict both outputs and explain why the original text remains unchanged.

````python
raw = "  Ada ; 18  "
name_text, age_text = raw.split(";")
name = name_text.strip()
age = int(age_text.strip())
print(f"{name} will be {age + 1} next year.")
print(raw)
````
