## 1. Name a procedure and reuse it

A function groups instructions under a name. **Parameters** name the inputs in the definition; **arguments** are the values supplied in a call. A **return value** lets the caller use the result.

The definition stores the procedure; the body runs when it is called. A docstring describes its purpose and assumptions.

````python
def triangle_area(base, height):
    """Return triangle area for positive base and height in the same unit."""
    area = base * height / 2
    return area

print(triangle_area(4, 10))
small_area = triangle_area(3, 5)
print(small_area)
````

## 2. Returning and printing do different jobs

`print` displays a value. `return` sends a value to the caller and ends the current function call. A function that reaches its end without an explicit return value returns `None`.

Predict the lines displayed below and the values stored in `shown` and `returned`.

````python
def show_double(number):
    print(number * 2)

def double(number):
    return number * 2

shown = show_double(4)
returned = double(4)
print("shown:", shown)
print("returned:", returned)
````

## 3. Local scope

Parameters and names assigned inside a function are normally local to that call. Reusing the same name outside the function does not make it the same variable.

Predict both outputs. The value passed in is 5; the outside variable happens to be called `number` too. For our numeric examples, changing the local variable does not change the caller’s variable. We will discuss shared mutable lists in the next notebook.

````python
def add_one(number):
    number = number + 1
    return number

number = 10
print(add_one(5))
print(number)
````

## Function contracts and early returns

**Core.** A contract states allowed inputs and the meaning of the result. A default argument supplies a value when a caller omits it; a keyword argument names the parameter explicitly. Use an early return when a case is already decided.

Assume integer scores and thresholds between 0 and 20. Predict all three calls.

````python
def passed(score, threshold=10):
    """Return whether score meets threshold; both are integers from 0 to 20."""
    return score >= threshold

print(passed(10))
print(passed(10, threshold=12))
print(passed(score=12, threshold=12))
````

## Check a contract with assertions

An assertion checks a condition and raises `AssertionError` when it is false. Use assertions to record expected results during development; they are not a substitute for validating user input. A successful assertion displays nothing.

A test needs an independently chosen expected answer. Testing only one ordinary input can miss a boundary error.

````python
assert passed(9) is False
assert passed(10) is True
assert passed(20) is True
assert passed(11, threshold=12) is False
print("Pass/fail checks passed")
````
