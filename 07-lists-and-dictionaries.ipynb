## 1. Lists store an ordered collection

A list can contain multiple values, including duplicates. Use square brackets to create one. Indexing and slicing work as they do for strings, but lists are **mutable**: their elements can be changed.

Predict the contents after each operation. `append` adds at the end; `pop` removes and returns an element, by default the last.

```python
scores = [4, 7, 4]
print(len(scores))
print(scores[0])
scores[1] = 8
scores.append(10)
print(scores)
removed = scores.pop()
print(removed)
print(scores)
```

## 2. Worked example — Sum a list

**Input:** a list of numbers, possibly empty. **Output:** their total. The empty-list total is 0. The function leaves the list unchanged.

Compare this with summing a range: only the source of the values has changed.

```python
def total_of(numbers):
    """Return the total of a list of numbers without changing the list."""
    total = 0
    for number in numbers:
        total = total + number
    return total

print(total_of([3, -1, 5]))
print(total_of([]))
```

## 3. Dictionaries map keys to values

A dictionary stores associations such as a student name and a score. Keys are unique; assigning to an existing key replaces its value. `key in dictionary` checks whether a key is present.

Use `dictionary[key]` when the key is known to exist. `dictionary.get(key, default)` gives a chosen default when it is absent.

```python
scores = {"Ada": 8, "Sam": 6}
print(scores["Ada"])
scores["Sam"] = 9
scores["Lee"] = 7
print(scores)
print("Mia" in scores)
print(scores.get("Mia", 0))
for name, score in scores.items():
    print(name, score)
```

## 4. Worked example — Word frequencies

**Input:** a string. **Output:** a dictionary mapping words to occurrence counts. For this task, convert to lowercase and split on whitespace; punctuation stays attached to words. Empty or whitespace-only text gives an empty dictionary.

`split()` produces a list of words. Each iteration either starts a word’s count or increments an existing count. Trace both occurrences of `red`.

```python
text = "Red blue red"
counts = {}
for word in text.lower().split():
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
```

**Alternative implementation:** replace the branch with `counts[word] = counts.get(word, 0) + 1`. Explain what the default 0 does before running it.

```python
# Trace and adapt the frequency-tabletext = "Red blue red"
counts = {}
for word in text.lower().split():
    counts[word] = counts.get(word, 0) + 1
print(counts) 
```

## List comprehensions — build a new collection

**Core demonstration.** A comprehension is a compact way to build a new list. First understand the equivalent loop, then compare the two forms.

```python
numbers = [1, 2, 3, 4]
even_numbers = []
for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
print(even_numbers)

same_result = [number for number in numbers if number % 2 == 0]
print(same_result)
```

## Structured data — a list of records

**Core.** A list preserves a sequence of students; each dictionary groups the fields belonging to one student. A nested list stores that student’s scores. We will use the same model in the workshop and turn it into classes next session.

`enumerate` supplies an index and a value. Read nested structures one level at a time.

```python
students = [
    {"name": "Ada", "scores": [12, 18]},
    {"name": "Sam", "scores": [8]},
    {"name": "Lee", "scores": []},
]
for position, student in enumerate(students, start=1):
    print(position, student["name"], student["scores"])

names = [student["name"] for student in students]
print(names)
```

## Nested mutation — a shallow copy still shares inner objects

Predict the output. A shallow copy creates a new outer list but keeps references to the dictionaries inside it. Mutating a nested scores list is visible through both outer lists. By contrast, a comprehension of immutable name strings does not expose the score lists.

When designing a function or a class, decide who owns mutable data and whether callers may modify it.

```python
records = [{"name": "Ada", "scores": [12]}]
outer_copy = records.copy()
outer_copy[0]["scores"].append(18)
print(records)
print(outer_copy is records)
print(outer_copy[0] is records[0])
```
